# **Implementation**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Deeplearning/datasets/placements.csv")
```

```python
print(df.shape)
df.head()
```

```python
sns.scatterplot(x=df['cgpa'], y=df['resume_score'], hue=df['placed'])
```

```python
X = df.iloc[:,0:2]
y = df.iloc[:, -1]
```

```python
from sklearn.linear_model import Perceptron
```

```python
p = Perceptron(
    max_iter=10000,
    tol=None,
    random_state=42,
    shuffle=True
)
```

```python
p.fit(X, y)
```

```python
p.coef_                   # returns the value of w1 and w2
```

```python
p.intercept_              # value of bias (b)
```

```python
plot_decision_regions(X.values, y.values, clf=p, legend=2)
```
